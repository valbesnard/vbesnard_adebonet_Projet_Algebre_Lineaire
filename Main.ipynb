#### c) Démonstration par récurrence

__Initialisation__

Soit t = 0 :

$$v_0=A^0v_0$$
$$\Leftrightarrow v_0=Iv_0$$
$$\Leftrightarrow v_0=v_0$$

La proposition est vraie pour t = 0.

__Hérédité__

Supposons que la proposition est vraie pour un t >= 0, c'est à dire :

$$v_t=A^tv_0$$

D'après l'exervice précédent :

$$v_{t+1}=Av_t$$
$$\Leftrightarrow v_{t+1}=A (A^tv_0)$$
$$\Leftrightarrow v_{t+1}=A^{t+1}v_0$$

La proposition est donc héréditaire.

__Conclusion__

La priorié est vraie pour t = 0 et elle est héréditaire. Par le principe de récurrence, on a :

$$\forall t \in \mathbb{N}, v_t=A^tv_0$$

```python
#II.1.a)
A=np.array([[1/2,1/6,1/6],
           [1/4,2/3,1/6],
           [1/4,1/6,2/3]])

print(A)
print("taille = ", np.shape(A))
print("type = " ,type(A))
```

```python
#II.1.b)
def is_stochastic(A,string, e ) :    

    #Vérifier que la matrice est carrée
    if np.shape(A)[0] != np.shape(A)[1] :
        return(False)

    #Vérifier que les coefs sont positifs
    elif np.min(A) < -e :  #Si le min est <-e alors il existe au moins une valeur négative (à la tolérance près)
        return(False)

    #Vérifier qu'elle est stochastique par colonnes ou par lignes
    else :

        if string == "column" :
            x = np.sum(A, axis=0) - np.array([1,1,1])  
            return(np.max(abs(x)) < e)  
        #Si le vecteur x en valeur absolue a une coordonnée > e alors la somme de la colonne ne vaut pas 1 au seuil de tolérance près
        
        elif string == "line" :
            x = np.sum(A, axis=1) - np.array([1,1,1])
            return(np.max(abs(x)) < e)
        #Idem que dans le if précédent

        #Un message d'erreur si jamais le deuxième argument n'est ni line ni column
        else :
            return("erreur : il faut que le deuxième argument soit line ou column")
```

```python
#II.1.c)
#On définit une matrice B non carrée mais qui vérifie les autres propriétés
B=np.array([[1/2,1/6,1/6,1/2],
           [1/4,2/3,1/6,1/2],
           [1/4,1/6,2/3,1/4]])

#On définit une matrice C qui possède une valeur négative mais vérifie les autres propriétés
C=np.array([[-1/4,1/6,1/6],
           [1/2,2/3,1/6],
           [3/4,1/6,2/3]])


print("pour A :", is_stochastic(A,"column",1e-12))
print("pour B :", is_stochastic(B,"column",1e-12))
print("pour C :", is_stochastic(C,"column",1e-12))
print("pour A transposée :", is_stochastic(np.transpose(A),"column",1e-12))
print("pour A transposée sur les lignes:", is_stochastic(np.transpose(A),"line",1e-12))
```

On voit que l'algorithme fonctionne bien sur les 3 conditions que doivent respecter les matrices Stochastiques.
La fonction renvoie faux sur la transposée de A lorsque l'on laisse "column" en deuxième argument mais True si on le remplace par "line"

### 2. Simulation du système dynamique

```python
def simulate_markov(A,x0,T) : 
    
    res = [x0] #Variable résultat qui pour l'instant est simplement le vecteur x0. 
    #Les [] sont nécessaires pour indiquer que res est un array à 1 ligne et 3 colonnes

    x=x0  #Variable x qui nous servira pour la boucle for 

    #Vérification compatibilité dimension

    #On teste s'il y a une différence entre le nombre de colonnes de A et le nombre d'éléments de x0. 
    #S'il y en a une, il y a un problème de dimension et on arrête l'algorithme
    if np.shape(A)[1] != np.shape(x0)[0] :   
        return("erreur ; incompatibilité des dimensions")

    #Itérations

    else : 
        for k in range (1,T+1) : 
            x=np.matmul(A,x)                    #On calcule le x_k
            res = np.append(res, [x], axis=0)   #On ajoute la ligne à notre array res.
            
        return(res)     

v1=np.array([1/2,1/4,1/4])
v2=np.array([1,0,0])
v3=np.array([1/7,2/7,4/7])
v4=np.array([0,0,1])

print("test 1 :\n", simulate_markov(A,v1,15))
print("test 2 :\n", simulate_markov(A,v2,15))
print("test 3 :\n", simulate_markov(A,v3,15))
print("test 4 :\n", simulate_markov(A,v4,15))
```

### 6. Trois méthodes de calcul de $A^n$

#### Méthode 1 : Produit matriciel itératif

```python
def power_iterative(A,n) : 
    
    if n == 0 :
        return(np.identity(np.shape(A)[0]))
    
    else : 
        M=A  #Matrice M que l'on va utiliser dans l'itération
        for k in range (n-1) :
            M=M @ A
        return(M)

power_iterative(A,3)
```

#### Méthode 2 : Utilisation de la diagonalisation

```python
def power_diagonalized(P, D, n):
    #Il faut traiter le cas n=0 car sinon on aurait que la matrice D**0 renvoie une matrice carrée remplie de 1
    if n == 0 :
            return(np.identity(np.shape(D)[0]))
    else :
        return(P @ D**n @ np.linalg.inv(P))  #Comme la matrice est diagonale, on peut calculer D**n qui élève toutes les composantes à la puissance n

np.linalg.matrix_power(A,3)
```

#### Comparaison deux à deux

```python
#On définit une valeur Nmax
Nmax=100

#On créé 3 listes vides qui accueilleront les valeurs des écarts par la norme de Frobenius
L_it_diag, L_it_np, L_diag_np =[],[],[]

for i in range (Nmax) :
    L_it_diag.append(np.linalg.norm(power_iterative(A,i) - power_diagonalized(P, D, i)))
    L_it_np.append(np.linalg.norm(power_iterative(A,i) - np.linalg.matrix_power(A,i)))
    L_diag_np.append(np.linalg.norm(power_diagonalized(P, D, i) - np.linalg.matrix_power(A,i)))


#On définit une fonction qui prend une liste L et un seuil s et qui renvoie l'indice et la valeur de la première valeur inférieure au seuil
def val_inf_seuil(L,s) :
    #On va utiliser une boucle while
    x=L[0]
    i=0
    while(x>=s and i<len(L)-1) :
        i=i+1
        x=L[i]
    if i==len(L)-1 :
        return("Pas de valeur inférieure au seuil")
    else :
        return(x,i)


#On définit une fonction qui prend une liste L et un seuil s et qui renvoie l'indice et la valeur de la première valeur supérieure au seuil.
#Elle permettra de contrôler si l'écart reste bien inférieur au seuil fixé ensuite
def val_sup_seuil(L,s) :
    #On va utiliser une boucle while
    x=L[0]
    i=0
    while(x<=s and i<len(L)-1) :
        i=i+1
        x=L[i]
    if i==len(L)-1 :
        return("Pas de valeur supérieure au seuil")
    else :
        return(x,i)
```
