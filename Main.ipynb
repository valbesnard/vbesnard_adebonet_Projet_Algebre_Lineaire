#### c) Démonstration par récurrence

__Initialisation__

Soit t = 0 :

$$v_0=A^0v_0$$
$$\Leftrightarrow v_0=Iv_0$$
$$\Leftrightarrow v_0=v_0$$

La proposition est vraie pour t = 0.

__Hérédité__

Supposons que la proposition est vraie pour un t >= 0, c'est à dire :

$$v_t=A^tv_0$$

D'après l'exervice précédent :

$$v_{t+1}=Av_t$$
$$\Leftrightarrow v_{t+1}=A (A^tv_0)$$
$$\Leftrightarrow v_{t+1}=A^{t+1}v_0$$

La proposition est donc héréditaire.

__Conclusion__

La priorié est vraie pour t = 0 et elle est héréditaire. Par le principe de récurrence, on a :

$$\forall t \in \mathbb{N}, v_t=A^tv_0$$
