#### c) Démonstration par récurrence

__Initialisation__

Soit t = 0 :

$$v_0=A^0v_0$$
$$\Leftrightarrow v_0=Iv_0$$
$$\Leftrightarrow v_0=v_0$$

La proposition est vraie pour t = 0.

__Hérédité__

Supposons que la proposition est vraie pour un $t \geq 0$, c'est à dire :

$$v_t=A^tv_0$$

D'après l'exervice précédent :

$$v_{t+1}=Av_t$$
$$\Leftrightarrow v_{t+1}=A (A^tv_0)$$
$$\Leftrightarrow v_{t+1}=A^{t+1}v_0$$

La proposition est donc héréditaire.

__Conclusion__

La priorié est vraie pour t = 0 et elle est héréditaire. Par le principe de récurrence, on a :

$$\forall t \in \mathbb{N}, v_t=A^tv_0$$

```python
#II.1.a)
A=np.array([[1/2,1/6,1/6],
           [1/4,2/3,1/6],
           [1/4,1/6,2/3]])

print(A)
print("taille = ", np.shape(A))
print("type = " ,type(A))
```

```python
#II.1.b)
def is_stochastic(A,string, e ) :  
    """
    Entrées :
        A - Matrice carrée aux coefficients positifs.
        string - Méthode de vérification pour savoir si la matrice est stochastique.
        e - Seuil de tolérance.
    Sortie :
        Bolean - La matrice A est stochastic ?
    """

    #Vérifier que la matrice est carrée
    if np.shape(A)[0] != np.shape(A)[1] :
        return(False)

    #Vérifier que les coefs sont positifs
    elif np.min(A) < -e :  #Si le min est <-e alors il existe au moins une valeur négative (à la tolérance près)
        return(False)

    #Vérifier qu'elle est stochastique par colonnes ou par lignes
    else :

        if string == "column" :
            x = np.sum(A, axis=0) - np.array([1,1,1])  
            return(np.max(abs(x)) < e)  
        #Si le vecteur x en valeur absolue a une coordonnée > e alors la somme de la colonne ne vaut pas 1 au seuil de tolérance près
        
        elif string == "line" :
            x = np.sum(A, axis=1) - np.array([1,1,1])
            return(np.max(abs(x)) < e)
        #Idem que dans le if précédent

        #Un message d'erreur si jamais le deuxième argument n'est ni line ni column
        else :
            raise Exception("Il faut que le deuxième argument soit line ou column.")
```

```python
#II.1.c)
#On définit une matrice B non carrée mais qui vérifie les autres propriétés
B=np.array([[1/2,1/6,1/6,1/2],
           [1/4,2/3,1/6,1/2],
           [1/4,1/6,2/3,1/4]])

#On définit une matrice C qui possède une valeur négative mais vérifie les autres propriétés
C=np.array([[-1/4,1/6,1/6],
           [1/2,2/3,1/6],
           [3/4,1/6,2/3]])


print("pour A :", is_stochastic(A,"column",1e-12))
print("pour B :", is_stochastic(B,"column",1e-12))
print("pour C :", is_stochastic(C,"column",1e-12))
print("pour A transposée :", is_stochastic(np.transpose(A),"column",1e-12))
print("pour A transposée sur les lignes:", is_stochastic(np.transpose(A),"line",1e-12))
```

On voit que l'algorithme fonctionne bien sur les 3 conditions que doivent respecter les matrices Stochastiques.
La fonction renvoie faux sur la transposée de A lorsque l'on laisse "column" en deuxième argument mais True si on le remplace par "line"

### 2. Simulation du système dynamique

```python
def simulate_markov(A,x0,T) : 
    """
    Entrées :
        A - Matrice carré.
        x0 - Situation initial.
        T - Nombre d'itérations.
    Sortie :
        np.array - Vecteur de la situation après T itération.
    """
    
    res = [x0] #Variable résultat qui pour l'instant est simplement le vecteur x0. 
    #Les [] sont nécessaires pour indiquer que res est un array à 1 ligne et 3 colonnes

    x=x0  #Variable x qui nous servira pour la boucle for 

    #Vérification compatibilité dimension

    #On teste s'il y a une différence entre le nombre de colonnes de A et le nombre d'éléments de x0. 
    #S'il y en a une, il y a un problème de dimension et on arrête l'algorithme
    if np.shape(A)[1] != np.shape(x0)[0] :   
        raise Exception("Incompatibilité des dimensions, il faut shape(A)[1] = shape(A)[0].")

    #Itérations

    else : 
        for k in range (1,T+1) : 
            x=np.matmul(A,x)                    #On calcule le x_k
            res = np.append(res, [x], axis=0)   #On ajoute la ligne à notre array res.
            
        return(res)     

v1=np.array([1/2,1/4,1/4])
v2=np.array([1,0,0])
v3=np.array([1/7,2/7,4/7])
v4=np.array([0,0,1])

print("test 1 :\n", simulate_markov(A,v1,15))
print("test 2 :\n", simulate_markov(A,v2,15))
print("test 3 :\n", simulate_markov(A,v3,15))
print("test 4 :\n", simulate_markov(A,v4,15))
```

```python
#II.4.a)

#On définit la fonction. On fera attention car la fonction linalg.eig peut renvoyer des vecteurs imaginaires et valeur propres imaginaires.
def eigen_data(A) :
    """
    Entrée :
        A - Matrice.
    Sortie :
        np.array - Valeurs propre
        np.array - Vecteurs propre
    """
    eval, evect = np.linalg.eig(A)
    return(eval,evect)

eigen_data(A)
```

```python
def diagonalize (A,e) :
    """
    Entrées :
        A - Matrice A à diagonalisé.
        e - Tolérance.
    Sortie :
        np.array - Matrice de passage.
        np.array - Matrice diagonalisé.
    """
    P=eigen_data(A)[1]  #P prend la matrice de vecteurs propres associés à 

    #On vérifie que la matrice P est bien inversible (à e près car le calcul du det pourrais-être très proche de 0 mais non considéré nul)
    if abs(np.linalg.det(A))<e : 
        raise Exception("la matrice n'est pas inversible.")

    #Itération
    else : 
        D=np.diag(eigen_data(A)[0])

        #On n'oublie pas d'enlever les parties imaginaires dues aux erreurs d'arrondis
        P=np.real_if_close(P)
        D=np.real_if_close(D)

        #On renvoie les deux matrices avec P en premier et D en deuxième pour avoir le même ordre que lorsqu'on les multiplieras ensemble
        return(P,D)

P=diagonalize(A,1e-12)[0]
D=diagonalize(A,1e-12)[1]

print(P)
print(D)
print(np.allclose(A, P @ D @ np.linalg.inv(P), atol=1e-10))
```

#### Méthode 1 : Produit matriciel itératif

```python
def power_iterative(A,n) : 
    """
    Entrées :
        A - Matrice carré.
        n - Nombre d'itérations.
    Sortie :
        np.array - Matrice A puissance n.
    """
    
    if n == 0 :
        return(np.identity(np.shape(A)[0]))
    
    else : 
        M=A  #Matrice M que l'on va utiliser dans l'itération
        for k in range (n-1) :
            M=M @ A
        return(M)

power_iterative(A,3)
```

#### Méthode 2 : Utilisation de la diagonalisation

```python
def power_diagonalized(P, D, n):
    """
    Entrées :
        P - Matrice de passage.
        D - Matrice diagonalisé.
        n - Nombre d'itérations.
    Sorties :
        np.array - Matrice A puissance n.
    """
    #Il faut traiter le cas n=0 car sinon on aurait que la matrice D**0 renvoie une matrice carrée remplie de 1
    if n == 0 :
            return(np.identity(np.shape(D)[0]))
    else :
        return(P @ D**n @ np.linalg.inv(P))  #Comme la matrice est diagonale, on peut calculer D**n qui élève toutes les composantes à la puissance n

power_diagonalized(P,D,3)
```

#### Comparaison deux à deux

```python
#On définit une valeur Nmax
Nmax=100

#On créé 3 listes vides qui accueilleront les valeurs des écarts par la norme de Frobenius
L_it_diag, L_it_np, L_diag_np =[],[],[]

for i in range (Nmax) :
    L_it_diag.append(np.linalg.norm(power_iterative(A,i) - power_diagonalized(P, D, i)))
    L_it_np.append(np.linalg.norm(power_iterative(A,i) - np.linalg.matrix_power(A,i)))
    L_diag_np.append(np.linalg.norm(power_diagonalized(P, D, i) - np.linalg.matrix_power(A,i)))


#On définit une fonction qui prend une liste L et un seuil s et qui renvoie l'indice et la valeur de la première valeur inférieure au seuil
def val_inf_seuil(L,s) :
    """
    Entrées :
        L - Liste.
        s - Seuil.
    Sorties :
        x - Première valeur inférieur au seuil.
        i - Indice de la valeur.
    """
    #On va utiliser une boucle while
    x=L[0]
    i=0
    while(x>=s and i<len(L)-1) :
        i=i+1
        x=L[i]
    if i==len(L)-1 :
        return(None, None)
    else :
        return(x,i)


#On définit une fonction qui prend une liste L et un seuil s et qui renvoie l'indice et la valeur de la première valeur supérieure au seuil.
#Elle permettra de contrôler si l'écart reste bien inférieur au seuil fixé ensuite
def val_sup_seuil(L,s) :
    """
    Entrées :
        L - Liste.
        s - Seuil.
    Sorties :
        x - Première valeur supérieur au seuil.
        i - Indice de la valeur.
    """
    #On va utiliser une boucle while
    x=L[0]
    i=0
    while(x<=s and i<len(L)-1) :
        i=i+1
        x=L[i]
    if i==len(L)-1 :
        return(None, None)
    else :
        return(x,i)
```

#### Preuve que la $\rho(A) \leq 1$ pour $A$ stochastique

Soient $A$ une matrice stochastique par colonne, $\lambda$ une valeur propre de $A$ et $x_0$ un vecteur propre associé à $\lambda$. On prend $x_0$ de sort que $|(\left\lVert x_0 \right\lVert_{1})=1$

On commence par définir une la norme matricielle suivante définit à partir de la norme 1 des vecteurs de $\mathbb{R}^n$:

$$
\left\lVert A \right\lVert_{1} = \sup_{\left\lVert x \right\lVert_{1} = 1}(\left\lVert Ax \right\lVert_{1})
$$

Pour tout $x \in \mathbb{R}^n$ on a que $ \left\lVert Ax \right\lVert_{1} \leq \left\lVert A \right\lVert_{1} \left\lVert Ax \right\lVert_{1} $

On a donc que 

$$
| \lambda | = | \lambda |\left\lVert x_0 \right\lVert_{1} = \left\lVert \lambda x_0 \right\lVert_{1} = \left\lVert Ax_0 \right\lVert_{1} \leq \left\lVert A \right\lVert_{1} \left\lVert x_0 \right\lVert_{1} = \left\lVert A \right\lVert_{1}
$$

Ainsi on a démontré que $| \lambda | \leq \left\lVert A \right\lVert_{1}$

Montrons maintenant que $\left\lVert A \right\lVert_{1} \leq 1$

On a pour tout $i \in \lbrace 1,...,n \rbrace $

$$
\begin{matrix}
\left\lVert Ae_i \right\lVert_{1} & = \sum_{k=1}^{n} |a_{k,i}| & \\
                                  & = \sum_{k=1}^{n} a_{k,i}   & \textrm {car les coefficients de A sont positifs} \\
                                  & = 1                        & \textrm{car A est stochastique par colonne}
\end{matrix}
$$

Soit maintenant un $x \in \mathbb{R}^n$ tel que $\left\lVert x \right\lVert_{1} =1$. 

On note $ \sum_{i=1}^{n} x_i e_i $ sa décomposition dans la base canonique.

$$
\begin{matrix}
\left\lVert Ax \right\lVert_{1} & = \left\lVert A \sum_{i=1}^{n} x_i e_i \right\lVert_{1} & \\
                                & = \left\lVert  \sum_{i=1}^{n} x_i Ae_i  \right\lVert_{1} & \\
                                & \leq \sum_{i=1}^{n} |x_i| \left\lVert A e_i  \right\lVert_{1} & \textrm{Or on a pour tout} i \in \lbrace 1,...,n \rbrace, \left\lVert A e_i  \right\lVert_{1} = 1 \textrm{ donc} \\
                                & \leq \sum_{i=1}^{n} |x_i| & \\ 
                                & \leq \left\lVert x \right\lVert_{1} & \\
                                & \leq 1
\end{matrix}
$$

Ainsi $\left\lVert A \right\lVert_{1} \leq 1$ et on a donc que $| \lambda | \leq \left\lVert A \right\lVert_{1} \leq 1 $

N.B. : Puisque tous les $Ae_i$ sont de norme 1, on a montré que $A$ était exactement de norme 1 



#### Concernant l'inégalité stricte ensuite

Une matrice stochastique est une matrice carrée à coefficient rééls positifs. Si ses coefficient sont même strictement positifs, elle rentre dans les hypothèses du théorème de Perron-Frobenius. 
Ainsi, l'espace propre associé à sa valeur propre maximale (qui pour nous est $\lambda = 1$) est de dimension 1. 
Tout autre vecteur propre, non colinéaire à $\pi$, de $A$ est donc associé à une valeur propre $\lambda '<1$

## Bonus 2 : Rôle de la deuxième valeur propre dominante

#### a) Pourquoi elle gouverne en général la vitesse asymptotique de convergence

Ici nous allons raisonner avec la norme 1. Comme toutes les normes sont équivalentes dans des espaces de Banach de dimension fini, nos résultats seront vrais, modulo un facteur, avec la norme de Frobenius ou toute autre norme.

Lorsque l'on s'intéresse à la vitesse de convergence de $v_t$ vers $\pi$ on s'intéresse à la convergence de $A^n$ vers $A_\infty$. 

En effet, on a

$$
\left\lVert v_t - \pi \right\lVert_{1} \\
=  \left\lVert A^tv_0 - A_\infty v_0 \right\lVert_{1} \\
=  \left\lVert (A^t - A_\infty )v_0 \right\lVert_{1} 
\leq \left\lVert A^t-A_\infty \right\lVert_{1} \left\lVert v_0 \right\lVert_{1} 
\leq \left\lVert A^t-A_\infty \right\lVert_{1} 
$$

car $\left\lVert v_0 \right\lVert_{1} = 1$

De plus, on a cette inégalité 

$$
\left\lVert A^t-A_\infty \right\lVert_{1} \\
=\left\lVert PD^nP^{-1}-PD_\infty P^{-1} \right\lVert_{1} \\
=\left\lVert PD^n - D_\infty P^{-1} \right\lVert_{1} \\
\leq \left\lVert P \right\lVert_{1} \left\lVert D^n - D_\infty \right\lVert_{1} \left\lVert P^{-1} \right\lVert_{1} \\
\leq (\left\lVert P \right\lVert_{1} \left\lVert P^{-1} \right\lVert_{1}) \left\lVert D^n - D_\infty \right\lVert_{1} 
$$

Où on a noté 
$$
D=diag(\lambda_1, \lambda_2 , \lambda_3) \\
D_\infty =diag(1,0,0)
$$
Avec $\lambda_1=1>\lambda_2 \geq \lambda_3$

De plus, on a pour toutes matrice $A$, on a que $\left\lVert A \right\lVert_{1} \leq \sup_{i,j=1,2,3}(|a_{i,j}|)$, ainsi $\left\lVert D^n - D_\infty \right\lVert_{1} \leq (\lambda_2)^n$ puisque c'est la plus grande valeur de $D^n - D_\infty$

Ainsi qu'il existe un $K>0$ tel que : 

$$
\left\lVert v_t - \pi \right\lVert_{1} \leq K(\lambda_2)^n
$$

Donc modulo une facteur $K>0$, la vitesse de convergence de la suite est celle de la deuxième valeur propre dominante.
