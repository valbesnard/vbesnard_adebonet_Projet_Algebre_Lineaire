### 1. Mise en équations du système

#### a) Ecriture des relations

$$x_{t+1} = \frac{1}{2}x_t + \frac{1}{6}y_t + \frac{1}{6}z_t$$
$$y_{t+1} = \frac{1}{4}x_t + \frac{2}{3}y_t + \frac{1}{6}z_t$$
$$z_{t+1} = \frac{1}{4}x_t + \frac{1}{6}y_t + \frac{2}{3}z_t$$

#### b) Ecriture matricielle

$$v_{t+1}=\begin{bmatrix} 
\frac{1}{2} & \frac{1}{6} & \frac{1}{6} \\
\frac{1}{4} & \frac{2}{3} & \frac{1}{6} \\
\frac{1}{4} & \frac{1}{6} & \frac{2}{3}
\end{bmatrix} \begin{pmatrix}x_t\\y_t\\z_t\end{pmatrix}$$

#### c) Démonstration par récurrence

__Initialisation__

Soit t = 0 :

$$v_0=A^0v_0$$
$$\Leftrightarrow v_0=Iv_0$$
$$\Leftrightarrow v_0=v_0$$

La proposition est vraie pour t = 0.

__Hérédité__

Supposons que la proposition est vraie pour un t >= 0, c'est à dire :

$$v_t=A^tv_0$$

D'après l'exervice précédent :

$$v_{t+1}=Av_t$$
$$\Leftrightarrow v_{t+1}=A (A^tv_0)$$
$$\Leftrightarrow v_{t+1}=A^{t+1}v_0$$

La proposition est donc héréditaire.

__Conclusion__

La priorié est vraie pour t = 0 et elle est héréditaire. Par le principe de récurrence, on a :

$$\forall t \in \mathbb{N}, v_t=A^tv_0$$

__Colonne 2 :__ $\frac{1}{6} + \frac{2}{3} + \frac{1}{6} = 1$

__Colonne 3 :__ $\frac{1}{6} + \frac{1}{6} + \frac{2}{3} = 1$

#### b) Vectuer de proportion

$v_t$ est un vecteur de proportion puisque : $x_t + y_t + z_t = 1$ et $x_t,y_t,z_t \geq 0$

Pour $v_{t+1}$ :

$$v_{t+1}=\begin{bmatrix} 
\frac{1}{2} & \frac{1}{6} & \frac{1}{6} \\
\frac{1}{4} & \frac{2}{3} & \frac{1}{6} \\
\frac{1}{4} & \frac{1}{6} & \frac{2}{3}
\end{bmatrix} \begin{pmatrix}x_t\\y_t\\z_t\end{pmatrix}$$

$$\Leftrightarrow v_{t+1}=
\begin{pmatrix} 
\frac{1}{2}x_t + \frac{1}{6}y_t + \frac{1}{6}z_t \\
\frac{1}{4}x_t + \frac{2}{3}y_t + \frac{1}{6}z_t \\
\frac{1}{4}x_t + \frac{1}{6}y_t + \frac{2}{3}z_t
\end{pmatrix}$$

D'après la question 1.a) :

$$\Leftrightarrow v_{t+1}=
\begin{pmatrix} 
x_{t+1} \\
y_{t+1} \\
z_{t+1}
\end{pmatrix}$$

Puisque $x_t,y_t,z_t \geq 0$ et les coefficients sont positifs alors $x_{t+1},y_{t+1},z_{t+1} \geq 0$.

$$x_{t+1} + y_{t+1} + z_{t+1} = \frac{1}{2}x_t + \frac{1}{6}y_t + \frac{1}{6}z_t + \frac{1}{4}x_t + \frac{2}{3}y_t + \frac{1}{6}z_t = \frac{1}{4}x_t + \frac{1}{6}y_t + \frac{2}{3}z_t$$

$$\Leftrightarrow x_{t+1} + y_{t+1} + z_{t+1} = x_t + y_t + z_t$$

$$\Leftrightarrow x_{t+1} + y_{t+1} + z_{t+1} = 1$$

En somme, $v_{t+1}$ respecte les conditions suivante : $\Leftrightarrow x_{t+1} + y_{t+1} + z_{t+1} = 1$ et $x_{t+1},y_{t+1},z_{t+1} \geq 0$ donc $v_{t+1}$ est un vecteur de proportion.

Dans notre situation, cela signifie que $\forall t \in \mathbb{N}, x_{t+1}, y_{t+1}, z_{t+1}$ resteront une proportion de population résident respectivement en zone urbaine, périurbaine et rurale.
